# Codando em *C*, *C++* e *R* com o ***Google Colab***
<p>Embora o foco principal do Google Colab seja programar em Python, podemos utilizar os "magic commands" do Jupyter para criar o arquivo .cpp e os comandos do terminal para compilar.</p>
<p>Isto é possível devido ao fato de o colab rodar sob uma máquina virtual Linux e já possuir os compiladores gcc e g++ instalados por padrão.</p>

No Jupyter Notebook, "magic commands" podem ser executados com `%` para comandos em uma única linha ou `%%` para comandos em toda a célula. Nos exemplos abaixo, o comando `%%writefile` cria o arquivo de texto (.c ou .cpp) com todo o conteúdo da célula.

## Exemplos em C

Segue abaixo exemplo do programa "Hello World" em C.

```python
%%writefile hello.c
#include <stdio.h>

int main() {
    printf("Hello, World!\n");
    return 0;
}
```

Após ter criado o arquivo "hello.c", iremos compilar usando o comando `!` para informar que a célula deve rodar no terminal. Iremos compilar com o `gcc`, já instalado por padrão na máquina virtual Linux.

```python
!gcc hello.c -o hello
```

E por fim, executamos o programa novamente com o comando `!`. Os comandos destes procedimentos são semelhantes a quando se programa diretamente em um terminal ou ambiente sem interface gráfica.

```python
!./hello
```

## Exemplos em C++

Para códigos em C++, o procedimento é bastante semelhante, a estrutura e sequência dos procedimentos é a mesma mudando apenas a linguagem, a extensão do arquivo e o compilador.

```python
%%writefile ola_mundo.cpp
#include <iostream>
using namespace std;

int main() {
    cout << "Olá do C++ no Colab!" << endl; //teste
    cout << "Este é um código escrito em C++." << endl;
    return 0;
}
```

Podemos executar comandos do shell adicionando uma exclamação precedendo o comando, no exemplo abaixo usamos o compilador `g++` para criar o programa `ola_mundo`.

```python
!g++ ola_mundo.cpp -o ola_mundo
```

Em seguida, executamos o programa utilizando o `./`. Note que não foi necessário utilizar o comando `chmod` para dar permissão de execução ao programa.

```python
!./ola_mundo
```

## Exemplos em R

Codar em R exige alguns procedimentos adicionais.

O primeiro passo é instalar a biblioteca `rpy2`.

```python
pip install rpy2
```

Este passo é apenas par aassegurar caso a biblioteca `rpy2` não esteja instalada. Agora podemos carregar a extensão `rpy2.ipython` e usar a palavra mágica `%%R` para executar código em R. Note que neste caso iremos fazer algo semelhante ao já demonstrado ao programar em C e C++.

```python
%load_ext rpy2.ipython
```

```python
%%R
x <- seq(-2*pi, 2*pi, length.out = 100)
y <- sin(x)
plot(x, y, type = "l", col = "blue", lwd = 2, main = "Gráfico da Função Seno", xlab = "x", ylab = "sin(x)")
```

O exemplo acima mostra a plotagem de um gráfico. Observe que o gráfico é plotado utilizando um modo gráfico de exibição de imagem, não necessitando de janela em pop-up ou outro elemento externo.

É possível ainda configurar o Google Colab para programar em **R** ou em **Julia*** clicando em _"Runtime"_ e _"Change runtime type"_. Isso é mais poderoso para projetos que exigem um ambiente R ou Julia completo com todos os seus pacotes e ferramentas nativas, diferentemente de usar `rpy2` para integrar R dentro de um notebook Python.
